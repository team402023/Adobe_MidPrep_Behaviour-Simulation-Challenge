```python
from transformers import BertTokenizer, BertModel
import torch

# Load pre-trained BERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
text = "Hi there this is Tharak here."
tokens = tokenizer(text, return_tensors='pt')
```

```python
with torch.no_grad():
    outputs = model(**tokens)

# Extract embeddings from the last layer
embeddings = outputs.last_hidden_state
```

```python
# Example of mean pooling
pooled_embeddings = torch.mean(embeddings, dim=1)
```

```python
def tokens(corpus):
  arr =[]
  for i in range(len(corpus)):
    text = corpus[i]
    tokens = tokenizer(text, return_tensors='pt')
    with torch.no_grad():
       outputs = model(**tokens)

    # Extract embeddings from the last layer
    embeddings = outputs.last_hidden_state
    # Example of mean pooling
    pooled_embeddings = torch.mean(embeddings, dim=1)
    arr.append(pooled_embeddings)
    return arr

```

```python
a = ["hi there this is tharak","this is a great place called","you are a great person","greatest of the greatest people"]
o = tokens(a)
type(o)
```
